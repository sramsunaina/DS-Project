
Key Hyperparameters to Know

• n_estimators: The number of decision trees built in the ensemble. Setting this between 100 and 300 usually gives stable results.
• max_depth: The maximum depth of each individual tree. Leaving it as None can still cause overfitting, so setting a limit (like 5 to 15) helps keep train and test accuracy close.
• min_samples_split: The minimum number of samples required to split an internal node. Increasing this from the default 2 to 5 or 10 reduces noise memorization.
• n_jobs=-1: Tells scikit-learn to use all available CPU cores to train the trees faster on larger datasets.

Why Switch from Decision Tree to Random Forest?

• Reduces Overfitting: A single decision tree memorizes noise easily, whereas a Random Forest averages predictions across many trees trained on random subsets of data and features.
• Higher Test Accuracy: Combining multiple diverse trees almost always improves generalization on unseen test data compared to a single tree.
